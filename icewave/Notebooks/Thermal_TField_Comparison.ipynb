```python
import matplotlib as mpl
def get_value_from_cm(color, cmap, colrange=[0.,1.]):
    color=mpl.colors.to_rgb(color/256)
    r = np.linspace(colrange[0],colrange[1], 256)
    norm = mpl.colors.Normalize(colrange[0],colrange[1])
    mapvals = cmap(norm(r))[:,:3]
    distance = np.sum((mapvals - color)**2, axis=1)
    return r[np.argmin(distance)]
```

```python
nx,ny,nc = d['im'].shape
IM = np.zeros((nx,ny))
for i in range(nx):
    for j in range(ny):
        IM[i,j] = get_value_from_cm(d['im'][i,j,:], cbar.cmap, colrange=[0.,1.])
```
