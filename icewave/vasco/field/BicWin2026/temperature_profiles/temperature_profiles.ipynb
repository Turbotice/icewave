```python
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
path_to_temperatures_cores_maddie = 'H:/data/temperatures/temperatures_cores_maddie'

lsdr = os.listdir(path_to_temperatures_cores_maddie)

Temperatures = {}
Depths_cores = {}

for name in lsdr:
    data = np.loadtxt(f'{path_to_temperatures_cores_maddie}/{name}', skiprows=1)
    Depths_cores[name] = data[:,0]
    Temperatures[name] = data[:,1]

```

```python
fig, ax = plt.subplots()
for name in lsdr:
    ax.plot(Temperatures[name], Depths_cores[name], 'o')
ax.invert_yaxis()

ax.set_xlabel('Temperature [°C]')
ax.set_ylabel('Ice Depth (0 at surface) [cm]')
```

```python
# fit the best one  with line (ICI ON ECHANGE LES AXES X ET Y)

name1 = lsdr[0]

ydata = Temperatures[name1]
xdata = Depths_cores[name1] * 1e-2

A, B = np.polyfit(xdata, ydata, 1)

plt.figure()
plt.plot(xdata, ydata, 'o')
plt.plot(xdata, xdata*A+B)
plt.ylabel('T')
plt.xlabel('Depth [m]')
plt.title('Fit : a='+str(A)+' , b='+str(B))
```
